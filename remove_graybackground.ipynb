```python
import cv2
import os
import numpy as np
import rembg
```

```python
wd = os.getcwd()
flds = {"trainingblock_1", "trainingblock_2", "trainingblock_3", "trainingblock_4"}

paths = []

for folder in flds:
    folder_path = os.path.join(wd, folder) 
    print(folder_path)
    if os.path.exists(folder_path) and os.path.isdir(folder_path):  # Check if the folder exists
        print(f"Files in {folder}:")
        for file in os.listdir(folder_path):  # List files in the folder
            file_path = os.path.join(folder_path, file)
            if os.path.isfile(file_path):  # Ensure it's a file
                print(file)
                paths.append(file_path) # Add the file path to the list
                
```

```python
session = rembg.new_session("isnet-general-use")
```

```python
for img in paths:
    print(img)
    base_name = os.path.basename(img)
    thepath = os.path.dirname(img)
    ext = os.path.splitext(base_name)[1]
    name = os.path.splitext(base_name)[0]
    
    #read one image
    im = cv2.imread(img)
    
    if im is None:
        print("Could not read the image.")

    
    img_nobg = rembg.remove(im, session = session,  alpha_matting=True, alpha_matting_foreground_threshold=270,alpha_matting_background_threshold=20, alpha_matting_erode_size=11, post_process_mask=True)

    #save the image
    cv2.imwrite(thepath + f"/{name}_nobg{ext}", img_nobg)
    print(thepath + f"/{name}_nobg{ext} saved")
    

  
```
